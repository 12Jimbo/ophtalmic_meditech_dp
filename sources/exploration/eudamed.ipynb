# EUDAMED API exploration

This notebook uses the public EUDAMED route. The API key is optional: the working public request uses `format=csv&api-version=v1.0` without authentication. If a deployment asks for a key, set `EUDAMED_API_KEY` without storing it in this notebook.

```python
import io
import os
import pandas as pd
import requests

BASE_URL = "https://api.datalake.sante.service.ec.europa.eu/eudamed"
API_KEY = os.environ.get("EUDAMED_API_KEY")
```

```python
def eudamed_get(endpoint, params=None, fmt="csv"):
    """Call an endpoint and parse JSON or CSV output."""
    query = dict(params or {})
    query["format"] = fmt
    query["api-version"] = "v1.0"
    headers = {"Content-Type": "application/json"}
    if API_KEY:
        headers["Ocp-Apim-Subscription-Key"] = API_KEY
    response = requests.get(
        f"{BASE_URL}/{endpoint.lstrip('/') }",
        headers=headers,
        params=query,
        timeout=30,
    )
    response.raise_for_status()
    return pd.read_csv(io.StringIO(response.text)) if fmt == "csv" else response.json()
```

## Sample queries

The filters are deliberately small and can be adjusted.

```python
actors = eudamed_get("actors", {"ACT_COUNTRY_ISO2_CODE": "IT"}, fmt="csv")
display(actors)
```

```python
reference = eudamed_get("reference", {"LANGUAGE": "en"}, fmt="csv")
display(reference)
```
